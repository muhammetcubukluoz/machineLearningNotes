```python
X = data.drop("selling_price", axis=1 )
y = data["selling_price"]
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=15)
```

```python
categoricalCol = data.select_dtypes("object").columns.to_list()
```

```python
categoricalCol
```

```python
uniqueValues = data[categoricalCol].nunique()
```

```python
uniqueValues
```

```python
# seller_type, fuel_type, transmission_type -> one hot encoding
# car_name, brand, model -> frequency encoding
```

```python
onehotCols = ["seller_type", "fuel_type", "transmission_type"]
frequencyCols = ["car_name", "brand", "model"]
```

```python
for col in frequencyCols:
    frequency = X_train[col].value_counts() / len(X_train)

    X_train[col + "_frequency"] = X_train[col].map(frequency)

    X_test[col + "_frequency"] = X_test[col].map(frequency)

    meanFrequency = frequency.mean()
    X_test[col + "_frequency"] = X_test[col+"_frequency"].fillna(meanFrequency)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
X_train = X_train.drop(["car_name", "brand", "model"], axis=1)
X_test = X_test.drop(["car_name", "brand", "model"], axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(
    transformers=[
        ("one hot", OneHotEncoder(handle_unknown="ignore", drop="first"), onehotCols)
    ], remainder="passthrough"
)
```

```python
X_train = transformer.fit_transform(X_train)
X_test = transformer.transform(X_test)
```

```python
encodedCols = transformer.get_feature_names_out()
```

```python
encodedCols
```

```python
X_train = pd.DataFrame(X_train, columns=encodedCols)
X_test = pd.DataFrame(X_test, columns=encodedCols)
```

```python
X_train.head()
```

```python
X_train.info()
```

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor

from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = AdaBoostRegressor()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("R2 Score:", r2_score(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
```

```python
params = {
    "n_estimators": [50, 80, 100, 120],
    "learning_rate": [0.001, 0.01, 0.1, 1, 2.0],
    "loss": ["linear", "square", "exponential"],
}
```

```python
rcv = RandomizedSearchCV(estimator=model, param_distributions=params, n_jobs=-1, cv=5, scoring="r2")
```

```python
rcv.fit(X_train, y_train)
```

```python
rcv.best_params_
```

```python
y_pred = rcv.predict(X_test)

print("R2 Score:", r2_score(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
```

```python
# https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.AdaBoostRegressor.html
# estimator parameter in DecionTreeRegressor
params = {
    "estimator__max_depth" : [3,4,5],
    "n_estimators": [50, 80, 100, 120],
    "learning_rate": [0.001, 0.01, 0.1, 1, 2.0],
    "loss": ["linear", "square", "exponential"]
}
```

```python
rcv = RandomizedSearchCV(estimator=AdaBoostRegressor(DecisionTreeRegressor()), param_distributions=params, n_jobs=-1, cv=5, scoring="r2")
```

```python
rcv.fit(X_train, y_train)
```

```python
rcv.best_params_
```

```python
y_pred = rcv.predict(X_test)

print("R2 Score:", r2_score(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
```
